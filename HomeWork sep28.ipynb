Operators in Python 

1. Arithmetic Operator:

Data Types
Int

```python
a = 10
b = 3

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
```

2.Float 

```python
a = 10.5
b = 2.5

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
```

3.String

```python
a = "Hello "
b = "Python"

print(a + b)
print(a * 3)
```

4.Complex

```python
a = 5 + 2j
b = 2 + 1j

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

2.Comparison Operators

1.int

```python
a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a >= b)
```

2.Float

```python
a = 10.5
b = 5.5

print(a == b)
print(a > b)
print(a < b)
print(a != b)
print(a >= b)

```

3.String

```python
a = "Roshni"
b = "Roshni"

print(a == b)
print(a != b)
```

```python
a = "Roshni"
b = "Sahani"

print(a==b)
print(a!=b)
```

4.Boolean

```python
a = True
b = False

print(a == b)
print(a != b)
```
